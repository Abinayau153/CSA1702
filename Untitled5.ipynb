```python
# Bird Flying using First-Order Logic

def CanFly(bird):
    flying_birds = ["Eagle", "Sparrow"]

    if bird in flying_birds:
        return True
    else:
        return False


bird = input("Enter bird name: ")

if CanFly(bird):
    print(bird, "can fly")
else:
    print(bird, "cannot fly")
```
