```python
# Towers of Hanoi - First Order Logic

disks = ["D1", "D2", "D3"]
towers = ["A", "B", "C"]

# Predicate: On(disk, tower)
def On(disk, tower):
    return (disk, tower)

# Initial state
initial = [
    On("D1", "A"),
    On("D2", "A"),
    On("D3", "A")
]

# Goal state
goal = [
    On("D1", "C"),
    On("D2", "C"),
    On("D3", "C")
]

print("Initial State:")
for x in initial:
    print(x[0], "is on", x[1])

print("\nGoal State:")
for x in goal:
    print(x[0], "is on", x[1])

print("\nMinimum moves:", 2 ** len(disks) - 1)
```
